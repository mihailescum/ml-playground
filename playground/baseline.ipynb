```python
from uci_gas_sensor_dataset import UCIGasSensorDataset

dataset = UCIGasSensorDataset()
dataset.load()
```

```python
from evaluation import get_splits, run

from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

logreg_model = make_pipeline(StandardScaler(), LogisticRegression(max_iter=5000))
splits = get_splits("batch1", dataset.df["batch"])
run(dataset.df[dataset.feature_columns].values, dataset.df["gas_name"].values, logreg_model, splits)
```

```python
from sklearn.neighbors import KNeighborsClassifier
knn_model = make_pipeline(StandardScaler(), KNeighborsClassifier(5))
splits = get_splits("batch1", dataset.df["batch"])
run(dataset.df[dataset.feature_columns].values, dataset.df["gas_name"].values, knn_model, splits)
```
