```python
import torch

class RNN():
    def __init__(self, num_inputs, num_hidden, num_outputs):    
        self.W_IH = torch.normal(0, 1, size=(num_inputs, num_hidden))
        self.W_HH = torch.normal(0, 1, size=(num_hidden, num_hidden))
        self.W_HQ = torch.normal(0, 1, size=(num_hidden, num_outputs))
        self.b_H = torch.normal(0, 1, size=(1, num_hidden))
        self.b_Q = torch.normal(0, 1, size=(1, num_outputs))

    def forward(self, X, hidden_state=None): # X = (X_t)_t with X_t = 
        if not hidden_state:
            hidden_state = torch.zeros((X.shape[1], self.W_IH.shape[1]))
        outputs = []
        for X_t in X:
            print(hidden_state)
            hidden_state = torch.relu(X_t @ self.W_IH + hidden_state @ self.W_HH + self.b_H)
            outputs.append(hidden_state @ self.W_HQ + self.b_Q)
        return outputs

    def __call__(self, X):
        return self.forward(X)
```

```python
batch_size, num_inputs, num_hidden, num_outputs, num_steps = 1, 3, 3, 4, 10

rnn = RNN(num_inputs, num_hidden, num_outputs)
X = torch.ones((num_steps, batch_size, num_inputs))
rnn(X)
```

```python
class RNNTorch(torch.nn.Module):
    def __init__(self, num_inputs, num_hidden, num_outputs):
        super().__init__()
        self.rnn = torch.nn.RNN(num_inputs, num_hidden, num_layers=1, nonlinearity="relu")
        self.output_layer = torch.nn.Linear(num_hidden, num_outputs)

    def forward(self, X): # X = (X_t)_t with X_t = 
            rnn_outputs, _ = self.rnn(X)
            return self.output_layer(rnn_outputs)
```

```python
rnn_torch = RNNTorch(num_inputs, num_hidden, num_outputs)
X = torch.ones((num_steps, batch_size, num_inputs))
rnn_torch(X)
```
