```python
import os
from dotenv import load_dotenv
import tabpfn_client

load_dotenv()
tabpfn_client.set_access_token(os.getenv("API_TOKEN_PRIOR_LABS"))
```

```python
from uci_gas_sensor_dataset import UCIGasSensorDataset

dataset = UCIGasSensorDataset()
dataset.load()
```

```python
from evaluation import get_splits, run

from sklearn.pipeline import make_pipeline
import xgboost as xgb

xgb_model = make_pipeline(xgb.XGBClassifier())
splits = get_splits("batch1", dataset.df["batch"])
run(dataset.df[dataset.feature_columns].values, dataset.df["gas"].values, xgb_model, splits)
```

```python
from evaluation import get_splits, run

from sklearn.pipeline import make_pipeline
from tabpfn_client import TabPFNClassifier

tabpfn_model = make_pipeline(TabPFNClassifier())
splits = get_splits("batch1", dataset.df["batch"])
run(dataset.df[dataset.feature_columns].values, dataset.df["gas"].values, tabpfn_model, splits)
```
